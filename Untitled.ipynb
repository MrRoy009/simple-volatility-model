```python
import pandas as pd
import numpy as np

# =========================================
# 1. LOAD DATA (FROM UPLOADED FILES)
# =========================================
nifty = pd.read_csv("nifty.csv")
vix = pd.read_csv("india_vix.csv")

# clean column names
nifty.columns = nifty.columns.str.strip()
vix.columns = vix.columns.str.strip()

# parse dates safely
nifty["Date"] = pd.to_datetime(nifty["Date"], dayfirst=True)
vix["Date"]   = pd.to_datetime(vix["Date"], dayfirst=True)

# set index
nifty.set_index("Date", inplace=True)
vix.set_index("Date", inplace=True)



# -----------------------------------------
# 4. MERGE DATA
# -----------------------------------------
df = pd.DataFrame({
    "Close": nifty["Close"],
    "India_VIX": vix["Close"]
})

df.dropna(inplace=True)

# =========================================
# 5. REALIZED VOLATILITY FUNCTION
# =========================================
def realized_volatility(price, window):
    log_returns = np.log(price / price.shift(1))
    rv = log_returns.rolling(window).std() * np.sqrt(252)
    return rv * 100  # annualized %

# -----------------------------------------
# 6. COMPUTE RVs
# -----------------------------------------
df["RV_5"] = realized_volatility(df["Close"], 5)
df["RV_10"] = realized_volatility(df["Close"], 10)
df["RV_20"] = realized_volatility(df["Close"], 20)

# -----------------------------------------
# 7. VOLATILITY SPREAD
# -----------------------------------------
df["Vol_Spread"] = df["India_VIX"] - df["RV_10"]

# =========================================
# 8. SIGNAL LOGIC (DESK-GRADE)
# =========================================
df["Signal"] = "NO TRADE"

sell_vol = (
    (df["Vol_Spread"] > 3) &
    (df["RV_5"] < df["RV_10"]) &
    (df["RV_10"] < df["RV_20"])
)

buy_vol = (
    (df["Vol_Spread"] < -2) &
    (df["RV_5"] > df["RV_10"])
)

df.loc[sell_vol, "Signal"] = "SELL VOL"
df.loc[buy_vol, "Signal"] = "BUY VOL"

# -----------------------------------------
# 9. OUTPUT RESULTS
# -----------------------------------------
signals = df[df["Signal"] != "NO TRADE"][
    ["Close", "India_VIX", "RV_5", "RV_10", "RV_20", "Vol_Spread", "Signal"]
]

print("\n==== VOLATILITY SIGNALS (1 YEAR DATA) ====\n")
print(signals)

# =========================================
# TOMORROW'S SIGNAL
# =========================================

latest = df.iloc[-1]

print("\n==== SIGNAL FOR NEXT TRADING DAY ====\n")
print(f"Date used for signal: {df.index[-1].date()}")
print(f"Tomorrow's Signal: {latest['Signal']}")
print(f"Vol Spread: {latest['Vol_Spread']:.2f}")
print(f"IV: {latest['India_VIX']:.2f}")
print(f"RV(10): {latest['RV_10']:.2f}")
```
